```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
import ast
from matplotlib.gridspec import GridSpec

# Parse the data from the text format
# Create separate dataframes for each problem difficulty
easy_data = {
    'LLMs': ['Deepseek', 'GPT', 'Gemini', 'Human'],
    'Execution_Time': [1.2012, 1.1752, 1.2051, 1.0652],
    'passed_count': [9, 9, 9, 10],
    'failed_test_problems': [['countNegatives'], ['countNegatives'], ['countNegatives'], []]
}

medium_data = {
    'LLMs': ['Deepseek', 'GPT', 'Gemini', 'Human'],
    'Execution_Time': [0.1232, 0.1184, 0.1209, 0.1177],
    'passed_count': [8, 8, 8, 10],
    'failed_test_problems': [['searchMatrix', 'peakIndexInMountainArray'], 
                             ['searchMatrix', 'peakIndexInMountainArray'],
                             ['searchMatrix', 'peakIndexInMountainArray'], 
                             []]
}

hard_data = {
    'LLMs': ['Deepseek', 'GPT', 'Gemini', 'Human'],
    'Execution_Time': [0.095, 0.0935, 0.09, 0.0842],
    'passed_count': [7, 7, 8, 10],
    'failed_test_problems': [['minmaxGasDist', 'minOperations', 'query'], 
                             ['minmaxGasDist', 'minOperations', 'query'],
                             ['minmaxGasDist', 'query'], 
                             []]
}

df_easy = pd.DataFrame(easy_data)
df_medium = pd.DataFrame(medium_data)
df_hard = pd.DataFrame(hard_data)

# Calculate metrics for chart
for df in [df_easy, df_medium, df_hard]:
    df['failed_count'] = df['failed_test_problems'].apply(len)
    df['total_problems'] = df['passed_count'] + df['failed_count']
    df['success_rate'] = df['passed_count'] / df['total_problems'] * 100

# Set up colors for consistency
# model_colors = {
#     'Deepseek': '#3498db',  # Blue
#     'GPT': '#e74c3c',       # Red
#     'Gemini': '#2ecc71',    # Green
#     'Human': '#9b59b6'      # Purple
# }

# Create a figure with subplots
plt.figure(figsize=(18, 12))
gs = GridSpec(2, 3, height_ratios=[1, 1.2])

# 1. Bar Chart showing passed_count across all difficulty levels
ax1 = plt.subplot(gs[0, :])
bar_width = 0.2
x = np.arange(len(df_easy['LLMs']))

# Plot bars for each difficulty level
for i, (df, label, offset) in enumerate(zip(
    [df_easy, df_medium, df_hard],
    ['Easy', 'Medium', 'Hard'],
    [-bar_width, 0, bar_width]
)):
    bars = ax1.bar(x + offset, df['passed_count'], width=bar_width, 
                   label=f'{label} Problems', alpha=0.8)
    
    # Add value labels on the bars
    for bar in bars:
        height = bar.get_height()
        ax1.text(bar.get_x() + bar.get_width()/2., height + 0.1,
                 f'{height}', ha='center', va='bottom', fontsize=9)

ax1.set_title('Number of Passed Problems by Difficulty Level', fontsize=16)
ax1.set_ylabel('Passed Count', fontsize=14)
ax1.set_xticks(x)
ax1.set_xticklabels(df_easy['LLMs'])
ax1.legend(loc='upper center', bbox_to_anchor=(0.5, -0.05), ncol=3)
ax1.set_ylim(0, 11)  # Set y-axis limit to slightly above max value
ax1.grid(axis='y', linestyle='--', alpha=0.7)

# 2. Stacked bar chart for passed vs failed for each difficulty level
difficulty_labels = ['Easy', 'Medium', 'Hard']
subplot_positions = [(1, 0), (1, 1), (1, 2)]

for i, (df, label, pos) in enumerate(zip(
    [df_easy, df_medium, df_hard],
    difficulty_labels,
    subplot_positions
)):
    ax = plt.subplot(gs[pos])
    
    # Create stacked bars
    bottom_bars = ax.bar(df['LLMs'], df['passed_count'], label='Passed', 
                         color='#2ecc71', alpha=0.8)
    top_bars = ax.bar(df['LLMs'], df['failed_count'], bottom=df['passed_count'], 
                      label='Failed', color='#e74c3c', alpha=0.8)
    
    # Add success rate as text
    for j, (p, f, r) in enumerate(zip(df['passed_count'], df['failed_count'], df['success_rate'])):
        total = p + f
        ax.text(j, total/2, f"{r:.1f}%", ha='center', va='center', 
                color='white', fontweight='bold', fontsize=12)
    
    # Add the failed problem names as annotations
    for j, (model, failed_list) in enumerate(zip(df['LLMs'], df['failed_test_problems'])):
        if failed_list:
            failed_str = ', '.join(failed_list)
            ax.annotate(f"Failed: {failed_str}", 
                       (j, df['passed_count'].iloc[j] + df['failed_count'].iloc[j] + 0.2),
                       ha='center', va='bottom', fontsize=8, rotation=0,
                       wrap=True)
    
    ax.set_title(f'{label} Problems: Passed vs Failed', fontsize=14)
    ax.set_ylabel('Count', fontsize=12)  # Increased font size for better readability
    ax.set_ylim(0, max(df['total_problems']) + 1.5)  # Add space for annotations
    
    # Only show the legend on the first subplot
    if i == 0:
        ax.legend(loc='upper center', bbox_to_anchor=(1.5, -0.15), ncol=2)

# Adjust layout
plt.tight_layout()
plt.subplots_adjust(hspace=0.4, wspace=0.3)

# Save the figure
plt.savefig('problem_solving_performance.png', dpi=300, bbox_inches='tight')

# Show the plot
plt.show()

# Create a table showing only failed test names for each model and difficulty level
print("\nFailed Tests by Model and Difficulty Level:")

# Create a failure summary table
failure_data = {
    'Model': [],
    'Easy_Failed': [],
    'Medium_Failed': [],
    'Hard_Failed': []
}

for model in df_easy['LLMs']:
    failure_data['Model'].append(model)
    
    # Add failed tests for each difficulty level
    for df, level in zip([df_easy, df_medium, df_hard], ['Easy', 'Medium', 'Hard']):
        model_data = df[df['LLMs'] == model].iloc[0]
        failed = model_data['failed_test_problems']
        failure_data[f'{level}_Failed'].append(', '.join(failed) if failed else 'None')

# Create the failure summary DataFrame
failure_df = pd.DataFrame(failure_data)

# Display the failure table
#print(failure_df.to_string(index=False))

# Create a figure for displaying the failure table
plt.figure(figsize=(12, 4))
ax = plt.subplot(111)
ax.axis('off')

# Create table
table = ax.table(
    cellText=failure_df.values,
    colLabels=failure_df.columns,
    loc='center',
    cellLoc='center',
    colWidths=[0.15, 0.25, 0.3, 0.3]
)

# Style the table
table.auto_set_font_size(False)
table.set_fontsize(10)
table.scale(1, 1.5)
for (i, j), cell in table.get_celld().items():
    if i == 0:  # Header row
        cell.set_text_props(weight='bold', color='white')
        cell.set_facecolor('#4472C4')
    elif j == 0:  # Model column
        cell.set_text_props(weight='bold')
        cell.set_facecolor('#D9E1F2')
    else:
        cell.set_facecolor('#E9EDF4' if i % 2 == 0 else '#FFFFFF')

plt.tight_layout()
plt.savefig('failed_tests_table.png', dpi=300, bbox_inches='tight')
plt.show()
```

```python
import matplotlib.pyplot as plt
import numpy as np
# import ast # Use this if failed lists are consistently strings like "'[...]'""

# --- Comprehensive Data Structure ---
full_data = {
    'Deepseek': {'EASY': {'time': 1.2012, 'passed': 9, 'failed': ['countNegatives']},
                 'MEDIUM': {'time': 0.1232, 'passed': 8, 'failed': ['peakIndexInMountainArray']}, # Manually parsed list
                 'HARD': {'time': 0.0950, 'passed': 7, 'failed': ['minmaxGasDist', 'minOperations', 'query']}}, # Manually parsed list
    'GPT':      {'EASY': {'time': 1.1752, 'passed': 9, 'failed': ['countNegatives']},
                 'MEDIUM': {'time': 0.1184, 'passed': 8, 'failed': ['peakIndexInMountainArray']}, # Manually parsed list
                 'HARD': {'time': 0.0935, 'passed': 7, 'failed': ['minmaxGasDist', 'minOperations', 'query']}}, # Manually parsed list
    'Gemini':   {'EASY': {'time': 1.2051, 'passed': 9, 'failed': ['countNegatives']}, # Unchanged from previous example data
                 'MEDIUM': {'time': 0.1209, 'passed': 8, 'failed': ['peakIndexInMountainArray']}, # Manually parsed list
                 'HARD': {'time': 0.0900, 'passed': 8, 'failed': ['minmaxGasDist', 'query']}}, # Manually parsed list
    'Human':    {'EASY': {'time': 1.0652, 'passed': 10, 'failed': []},
                 'MEDIUM': {'time': 0.1177, 'passed': 9, 'failed': []},
                 'HARD': {'time': 0.0842, 'passed': 10, 'failed': []}}
}

# Define colors for each model (as provided previously)
model_colors = {
    'Deepseek': '#3498db',  # Blue
    'GPT': '#e74c3c',       # Red
    'Gemini': '#2ecc71',    # Green
    'Human': '#9b59b6'      # Purple
}

participants = list(full_data.keys())
# Use the keys from the data structure consistently
difficulty_keys = ['EASY', 'MEDIUM', 'HARD']
difficulty_labels = ['Easy', 'Medium', 'Hard'] # Cleaner labels for chart axis

n_participants = len(participants)
n_difficulties = len(difficulty_keys)

# --- Plotting Setup ---
# Create a single subplot for the chart
fig, ax = plt.subplots(figsize=(14, 8)) # Adjusted figure size for single plot

# --- Chart Plotting (on ax) ---
x = np.arange(n_difficulties)  # the label locations (centers of groups)
bar_width = 0.18  # Width of each bar
total_width = bar_width * n_participants # Total width for a group of bars

# Store bar positions for line plotting later
bar_positions = {participant: [] for participant in participants}

# Plot Bars and Add Text Labels
for i, participant in enumerate(participants):
    # Extract execution times
    times = [full_data[participant][diff_key]['time'] for diff_key in difficulty_keys]
    current_participant_positions = []
    for j in range(n_difficulties):
        position = x[j] - (total_width / 2) + (i * bar_width) + (bar_width / 2)
        current_participant_positions.append(position)

    bar_positions[participant] = current_participant_positions
    # Plot the bars
    bars = ax.bar(current_participant_positions, times, bar_width, label=participant, color=model_colors[participant])

    # Add text labels (Execution Time) above each bar
    for bar_index, bar in enumerate(bars):
        yval = bar.get_height()
        xval = bar.get_x() + bar.get_width() / 2
        label_text = f"{yval:.4f}"
        ax.text(xval, yval + 0.01, label_text,
                      ha='center',
                      va='bottom',
                      fontsize=12,
                      color='black',
                      rotation=0)

# Plot Lines (Overlay connecting bar tops)
line_styles = ['-', '--', ':', '-.']
markers = ['o', 's', '^', 'd']

for i, participant in enumerate(participants):
    times = [full_data[participant][diff_key]['time'] for diff_key in difficulty_keys]
    line_x_coords = bar_positions[participant]
    ax.plot(line_x_coords, times,
                  marker=markers[i % len(markers)],
                  linestyle=line_styles[i % len(line_styles)],
                  linewidth=2,
                  color=model_colors[participant],
                  label='_nolegend_')

# Chart Labels, Title, Ticks, Legend
ax.set_ylabel('Execution Time (seconds)')
ax.set_title('LLM vs Human: Execution Time on Coding Problems by Difficulty', pad=15)
ax.set_xticks(x)
ax.set_xticklabels(difficulty_labels, fontsize=14)
# Adjust y-limit slightly to ensure labels fit
ax.set_ylim(bottom=0, top=ax.get_ylim()[1] * 1.1) # Increase top limit by 10%
ax.legend(title="Participants")
ax.yaxis.grid(True, linestyle='--', alpha=0.7)


# --- Final Layout Adjustment & Display ---
plt.tight_layout(pad=2.0) # Add padding
plt.show()
```
