```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.gridspec import GridSpec

# Create the dataframe from the provided data
data = {
    'model': ['gpt-4-turbo', 'DeepSeek-V3', 'Gemini-2Flash', 'Human'],
    'total_MU': [74.94, 76.24, 74.09, 65.36],
    'normalized_MU': [1.04, 1.08, 1.05, 1.00]
}
df = pd.DataFrame(data)

# Set the style for the plots
plt.style.use('seaborn-v0_8-whitegrid')
colors = ['#8884d8', '#82ca9d']

# Create a figure with subplots
fig = plt.figure(figsize=(14, 10))
gs = GridSpec(3, 2, height_ratios=[4, 4, 3])

# Plot 1: Side-by-side total_MU
ax1 = fig.add_subplot(gs[0, 0])
bars1 = ax1.bar(df['model'], df['total_MU'], color=colors[0])
ax1.axhline(y=65.36, color='red', linestyle='--', label='Human Baseline (65.36)')
ax1.set_title('EffiBench Total MU by Model', fontsize=14)
ax1.set_xlabel('Model', fontsize=12)
ax1.set_ylabel('Total MU', fontsize=12)
ax1.set_ylim([60, 80])  # Adjusted for total_MU range
ax1.legend()
for bar in bars1:
    height = bar.get_height()
    ax1.text(bar.get_x() + bar.get_width()/2., height + 0.5,
             f'{height:.4f}', ha='center', va='bottom', fontsize=12)

# Plot 2: Side-by-side normalized_MU
ax2 = fig.add_subplot(gs[0, 1])
bars2 = ax2.bar(df['model'], df['normalized_MU'], color=colors[1])
ax2.axhline(y=1.00, color='red', linestyle='--', label='Human Baseline (1.00)')
ax2.set_title('EffiBench Normalized MU by Model', fontsize=14)
ax2.set_xlabel('Model', fontsize=12)
ax2.set_ylabel('Normalized MU', fontsize=12)
ax2.set_ylim([0.95, 1.15])  # Adjusted for normalized_MU range
ax2.legend()
for bar in bars2:
    height = bar.get_height()
    ax2.text(bar.get_x() + bar.get_width()/2., height + 0.005,
             f'{height:.2f}', ha='center', va='bottom', fontsize=12)

# Plot 3: Combined chart with dual y-axes
ax3 = fig.add_subplot(gs[1, :])
x = np.arange(len(df['model']))
width = 0.35

# Primary y-axis (total_MU)
bars3_1 = ax3.bar(x - width/2, df['total_MU'], width, label='Total MU', color=colors[0])
ax3.axhline(y=65.36, color='red', linestyle='--', label='Human Baseline (Total MU: 65.36)')
ax3.set_ylabel('Total MU', fontsize=12, color=colors[0])
ax3.tick_params(axis='y', labelcolor=colors[0])
ax3.set_ylim([60, 80])  # Range for total_MU

# Secondary y-axis (normalized_MU)
ax3_right = ax3.twinx()
bars3_2 = ax3_right.bar(x + width/2, df['normalized_MU'], width, label='Normalized MU', color=colors[1])
ax3_right.axhline(y=1.00, color='purple', linestyle='--', label='Human Baseline (Normalized MU: 1.00)')
ax3_right.set_ylabel('Normalized MU', fontsize=12, color=colors[1])
ax3_right.tick_params(axis='y', labelcolor=colors[1])
ax3_right.set_ylim([0.95, 2.15])  # Range for normalized_MU

# Common settings for Plot 3
ax3.set_title('EffiBench MU Metrics Comparison', fontsize=14)
ax3.set_xlabel('Model', fontsize=12)
ax3.set_xticks(x)
ax3.set_xticklabels(df['model'])

# Combined legend
lines1, labels1 = ax3.get_legend_handles_labels()
lines2, labels2 = ax3_right.get_legend_handles_labels()
ax3.legend(lines1 + lines2, labels1 + labels2, loc='upper right')

# Add value labels to bars
for bar in bars3_1:  # total_MU bars
    height = bar.get_height()
    ax3.text(bar.get_x() + bar.get_width()/2., height + 0.5,
             f'{height:.4f}', ha='center', va='bottom', fontsize=12)
for bar in bars3_2:  # normalized_MU bars
    height = bar.get_height()
    ax3_right.text(bar.get_x() + bar.get_width()/2., height + 0.005,
                   f'{height:.4f}', ha='center', va='bottom', fontsize=12)

# Create a table at the bottom
ax4 = fig.add_subplot(gs[2, :])
ax4.axis('off')  # Hide the axis
table_data = [df['model'], df['total_MU'], df['normalized_MU']]
table_rows = ['Model', 'Total MU', 'Normalized MU']
table = ax4.table(
    cellText=table_data,
    rowLabels=table_rows,
    loc='center',
    cellLoc='center',
    colWidths=[0.25] * len(df)
)
table.auto_set_font_size(False)
table.set_fontsize(12)
table.scale(1, 1.5)

# Adjust layout
plt.tight_layout()
plt.subplots_adjust(hspace=0.4)

# Save the figure
plt.savefig('ET_charts_original_with_baseline_dual_axis.png', dpi=300, bbox_inches='tight')

# Show the figure
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np
from matplotlib.gridspec import GridSpec

# Use the comprehensive data structure
full_data = {
    'Deepseek': {'EASY': {'time': 1.2012, 'passed': 9, 'failed': ['countNegatives']},
                 'MEDIUM': {'time': 0.1232, 'passed': 8, 'failed': ['peakIndexInMountainArray']},
                 'HARD': {'time': 0.0950, 'passed': 7, 'failed': ['minmaxGasDist', 'minOperations', 'query']}},
    'GPT':      {'EASY': {'time': 1.1752, 'passed': 9, 'failed': ['countNegatives']},
                 'MEDIUM': {'time': 0.1184, 'passed': 8, 'failed': ['peakIndexInMountainArray']},
                 'HARD': {'time': 0.0935, 'passed': 7, 'failed': ['minmaxGasDist', 'minOperations', 'query']}},
    'Gemini':   {'EASY': {'time': 1.2051, 'passed': 9, 'failed': ['countNegatives']},
                 'MEDIUM': {'time': 0.1209, 'passed': 8, 'failed': ['peakIndexInMountainArray']},
                 'HARD': {'time': 0.0900, 'passed': 8, 'failed': ['minmaxGasDist', 'query']}},
    'Human':    {'EASY': {'time': 1.0652, 'passed': 10, 'failed': []},
                 'MEDIUM': {'time': 0.1177, 'passed': 10, 'failed': []},
                 'HARD': {'time': 0.0842, 'passed': 10, 'failed': []}}
}

# Convert to DataFrame format for easier plotting
rows = []
for model, difficulties in full_data.items():
    for difficulty, stats in difficulties.items():
        # Calculate normalized time (Human as baseline)
        human_time = full_data['Human'][difficulty]['time']
        normalized_time = stats['time'] / human_time
        
        rows.append({
            'Model': model,
            'Difficulty': difficulty,
            'Execution_Time': stats['time'],
            'Normalized_Time': normalized_time,
            'Passed_Count': stats['passed'],
            'Failed_Count': len(stats['failed']),
            'Failed_Problems': ', '.join(stats['failed']) if stats['failed'] else 'None'
        })

df = pd.DataFrame(rows)

# Calculate average execution time and normalized time across all difficulties
avg_df = df.groupby('Model').agg({
    'Execution_Time': 'mean',
    'Normalized_Time': 'mean',
    'Passed_Count': 'mean'
}).reset_index()

# Set up colors for each model
model_colors = {
    'Deepseek': '#3498db',  # Blue
    'GPT': '#e74c3c',       # Red
    'Gemini': '#2ecc71',    # Green
    'Human': '#9b59b6'      # Purple
}

#ax1 = fig.add_subplot(gs[0, 0])
#ax2 = fig.add_subplot(gs[0, 1])
#ax3 = fig.add_subplot(gs[1, :])
# Create figure with subplots
fig = plt.figure(figsize=(15, 12))
gs = GridSpec(4, 2, height_ratios=[3, 3, 3, 1.5])

# 1. Bar chart: Execution Time by Model
ax1 = fig.add_subplot(gs[0, 0])
bars1 = []
for i, model in enumerate(avg_df['Model']):
    bar = ax1.bar(i, avg_df[avg_df['Model'] == model]['Execution_Time'].values[0],
                 color=model_colors[model], alpha=0.8)
    bars1.append(bar)

ax1.set_title('Raw Average Execution Time by Model', fontsize=14)
ax1.set_ylabel('Time (seconds)', fontsize=12)
ax1.set_xticks(range(len(avg_df)))
ax1.set_xticklabels(avg_df['Model'])
ax1.grid(axis='y', linestyle='--', alpha=0.7)

# Add value labels on bars
for i, model in enumerate(avg_df['Model']):
    height = avg_df[avg_df['Model'] == model]['Execution_Time'].values[0]
    ax1.text(i, height + 0.01, f'{height:.4f}', ha='center', va='bottom', fontsize=10)

# 2. Bar chart: Normalized Execution Time by Model
ax2 = fig.add_subplot(gs[0, 1])
bars2 = []
for i, model in enumerate(avg_df['Model']):
    bar = ax2.bar(i, avg_df[avg_df['Model'] == model]['Normalized_Time'].values[0],
                 color=model_colors[model], alpha=0.8)
    bars2.append(bar)

ax2.set_title('Raw Normalized Execution Time by Model', fontsize=14)
ax2.set_ylabel('Normalized Time (Human=1.0)', fontsize=12)
ax2.set_xticks(range(len(avg_df)))
ax2.set_xticklabels(avg_df['Model'])
ax2.axhline(y=1.0, color='gray', linestyle='--', alpha=0.7)
ax2.grid(axis='y', linestyle='--', alpha=0.7)

# Add value labels on bars
for i, model in enumerate(avg_df['Model']):
    height = avg_df[avg_df['Model'] == model]['Normalized_Time'].values[0]
    ax2.text(i, height + 0.01, f'{height:.3f}', ha='center', va='bottom', fontsize=10)

# 3. Bar chart: Execution Time Metrics Comparison with side-by-side bars
ax3 = fig.add_subplot(gs[1, :])
x = np.arange(len(avg_df['Model']))
width = 0.4

# Plotting both metrics side by side for each model
bar1 = ax3.bar(x - width/2, avg_df['Execution_Time'], width, label='Execution Time', color='#3498db')
bar2 = ax3.bar(x + width/2, avg_df['Normalized_Time'], width, label='Normalized Time', color='#e74c3c')

ax3.set_title('Raw Execution Time Metrics Comparison', fontsize=14)
ax3.set_ylabel('Values', fontsize=12)
ax3.set_xticks(x)
ax3.set_xticklabels(avg_df['Model'])
ax3.legend()
ax3.grid(axis='y', linestyle='--', alpha=0.7)

# Add value labels on bars
for i, bar in enumerate(bar1):
    height = bar.get_height()
    ax3.text(bar.get_x() + bar.get_width()/2, height + 0.01, 
             f'{height:.3f}', ha='center', va='bottom', fontsize=9)

for i, bar in enumerate(bar2):
    height = bar.get_height()
    ax3.text(bar.get_x() + bar.get_width()/2, height + 0.01, 
             f'{height:.3f}', ha='center', va='bottom', fontsize=9)

# 4. Detailed Table
ax4 = fig.add_subplot(gs[2, :])
ax4.axis('off')

# Create a more detailed table with difficulty-specific data
difficulties = ['EASY', 'MEDIUM', 'HARD', 'AVG']
models = avg_df['Model'].unique()

# Prepare table data
table_data = []
for model in models:
    model_row = [model]
    
    # Add raw execution times for each difficulty
    for difficulty in ['EASY', 'MEDIUM', 'HARD']:
        et = df[(df['Model'] == model) & (df['Difficulty'] == difficulty)]['Execution_Time'].values[0]
        model_row.append(f'{et:.4f}')
    
    # Add average execution time
    avg_et = avg_df[avg_df['Model'] == model]['Execution_Time'].values[0]
    model_row.append(f'{avg_et:.4f}')
    
    # Add normalized execution times for each difficulty
    for difficulty in ['EASY', 'MEDIUM', 'HARD']:
        norm_et = df[(df['Model'] == model) & (df['Difficulty'] == difficulty)]['Normalized_Time'].values[0]
        model_row.append(f'{norm_et:.3f}')
    
    # Add average normalized execution time
    avg_norm_et = avg_df[avg_df['Model'] == model]['Normalized_Time'].values[0]
    model_row.append(f'{avg_norm_et:.3f}')
    
    table_data.append(model_row)

# Column headers
col_labels = ['Model']
for diff in difficulties:
    col_labels.append(f'ET ({diff})')
for diff in difficulties:
    col_labels.append(f'Norm ET ({diff})')

# Create the table
table = ax4.table(
    cellText=table_data,
    colLabels=col_labels,
    loc='center',
    cellLoc='center'
)

# Style the table
table.auto_set_font_size(False)
table.set_fontsize(9)
table.scale(1, 1.5)
table.auto_set_column_width(col=list(range(len(col_labels))))

plt.tight_layout()
plt.subplots_adjust(hspace=0.4, wspace=0.2)

# Save the figure
plt.savefig('et_comparison_simple.png', dpi=120, bbox_inches='tight')
plt.show()
```
