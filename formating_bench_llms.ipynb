```python
from helper_functions import write_list_to_csv
import csv

data_bench = []
# read benchmark csv file get data first
with open("src/bench_llms.csv", 'r', newline='') as bech_data:
    reader = csv.reader(bech_data)
    for row in reader:
        data_bench.append(row[0][:-1].split("&"))
# header
header = ["model", "total_ET", "normalized_ET","total_MMU", "normalized_MMU", "total_MU", "normalized_MU"]

write_list_to_csv(header, "benchmark_llms.csv")
for bech_line in data_bench:
    write_list_to_csv(bech_line, "benchmark_llms.csv", True)
```
